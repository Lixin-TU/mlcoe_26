Get income statement and balance sheet data from yahoo finance.

Selected company: JP Morgan Chase & Co. (JPM).

In this notebook, we will retrieve income statement and balance sheet data for JPM from Yahoo Finance over the past 4 years using the `yfinance` library in Python.

```python
# ! pip install yfinance
```

```python
# import the liobrary and seeting the stock of interest
import yfinance as yf
import pandas as pd

# set the stock ticker symbol 
# Tencent - Listed on the Hong Kong Stock Exchange
Tencent = yf.Ticker('0700.HK')
# JP Morgan Chase - Listed on the New York Stock Exchange (NYSE)
JPM = yf.Ticker('JPM')
# Alibaba
Alibaba_HK = yf.Ticker('9988.HK') # Hong Kong Stock Exchange (HKEX)
# Exxon Mobil - Listed on the NYSE
Exxon = yf.Ticker('XOM')
# Volkswagen
VW_DE = yf.Ticker('VOW3.DE')  # XETRA (Germany).
# Microsoft
Microsoft = yf.Ticker('MSFT')
# Google (Alphabet) 
Google_A = yf.Ticker('GOOGL') # Class A shares (with voting rights)
# Google_C = yf.Ticker('GOOG')  # Class C shares (no voting rights)

ticker_symbol = Google_A
```
